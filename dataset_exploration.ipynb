```python
import os
import numpy as np

print("=" * 60)
print("OCEANEMBED DATA AUDIT")
print("=" * 60)

files = [
    "X_train.npy",
    "X_val.npy",
    "y_train.npy",
    "y_val.npy",
    "ocean_mask.npy",
    "input_scaler.npy",
    "best_model.pth",
    "val_predictions.npy",
    "ocean_mask_bool.npy"
]

print("\nFILES:")
for f in files:
    if os.path.exists(f):
        size_mb = os.path.getsize(f) / (1024**2)
        print(f"✓ {f:25s} {size_mb:.2f} MB")
    else:
        print(f"✗ {f:25s} MISSING")


# -----------------------------
# LOAD
# -----------------------------

Xtr = np.load("X_train.npy")
Xv  = np.load("X_val.npy")
ytr = np.load("y_train.npy")
yv  = np.load("y_val.npy")
mask_raw = np.load("ocean_mask.npy")

print("\nSHAPES:")
print("X_train:", Xtr.shape)
print("X_val  :", Xv.shape)
print("y_train:", ytr.shape)
print("y_val  :", yv.shape)
print("mask   :", mask_raw.shape)


# -----------------------------
# MASK
# -----------------------------

mask = mask_raw.astype(bool)

if mask.ndim == 3:
    mask = mask[0]

print("\nMASK:")
print("Shape:", mask.shape)
print("Ocean pixels:", mask.sum())
print("Total pixels:", mask.size)
print("Ocean fraction:", mask.mean())


# -----------------------------
# EXPECTED SHAPES
# -----------------------------

print("\nSHAPE CHECK:")

checks = {
    "X_train": Xtr.shape == (145, 76, 76, 3),
    "X_val":   Xv.shape  == (37, 76, 76, 3),
    "y_train": ytr.shape == (145, 76, 76, 14),
    "y_val":   yv.shape  == (37, 76, 76, 14),
    "mask":    mask.shape == (76, 76)
}

for name, ok in checks.items():
    print(("✓" if ok else "✗"), name, ok)


# -----------------------------
# NaN COUNTS
# -----------------------------

print("\nNAN COUNTS:")

for name, arr in [
    ("X_train", Xtr),
    ("X_val", Xv),
    ("y_train", ytr),
    ("y_val", yv)
]:
    print(f"{name:8s}: {np.isnan(arr).sum():,}")


# -----------------------------
# INPUT NaNs BY CHANNEL
# -----------------------------

print("\nINPUT NaNs BY CHANNEL:")

channel_names = ["SST", "SSH", "SSS"]

for i, name in enumerate(channel_names):
    train_nan = np.isnan(Xtr[..., i]).sum()
    val_nan   = np.isnan(Xv[..., i]).sum()

    print(
        f"{name}: "
        f"train={train_nan:,} "
        f"val={val_nan:,}"
    )


# -----------------------------
# OCEAN-ONLY INPUT NaNs
# -----------------------------

print("\nOCEAN-ONLY INPUT NaNs:")

for i, name in enumerate(channel_names):

    train_ocean = Xtr[..., i][:, mask]
    val_ocean   = Xv[..., i][:, mask]

    train_nan = np.isnan(train_ocean).sum()
    val_nan   = np.isnan(val_ocean).sum()

    train_total = train_ocean.size
    val_total   = val_ocean.size

    print(
        f"{name}: "
        f"train={train_nan/train_total:.4%} "
        f"val={val_nan/val_total:.4%}"
    )


# -----------------------------
# TARGET NaNs BY DEPTH
# -----------------------------

print("\nTARGET NaNs BY DEPTH:")

depths = np.array([
    0.5, 5, 10, 20, 30, 50, 75,
    100, 125, 150, 200, 300, 500, 700
])

for d, depth in enumerate(depths):

    train_vals = ytr[..., d][:, mask]
    val_vals   = yv[..., d][:, mask]

    train_nan = np.isnan(train_vals).mean()
    val_nan   = np.isnan(val_vals).mean()

    print(
        f"{depth:6.1f} m : "
        f"train={train_nan:.4%} "
        f"val={val_nan:.4%}"
    )


# -----------------------------
# TARGET RANGE
# -----------------------------

print("\nTARGET RANGE:")

print(
    "Train:",
    np.nanmin(ytr),
    "to",
    np.nanmax(ytr)
)

print(
    "Val  :",
    np.nanmin(yv),
    "to",
    np.nanmax(yv)
)


# -----------------------------
# TARGET OCEAN RANGE
# -----------------------------

print("\nTARGET OCEAN-ONLY RANGE:")

ocean_ytr = ytr[:, mask, :]
ocean_yv  = yv[:, mask, :]

print(
    "Train:",
    np.nanmin(ocean_ytr),
    "to",
    np.nanmax(ocean_ytr)
)

print(
    "Val:",
    np.nanmin(ocean_yv),
    "to",
    np.nanmax(ocean_yv)
)


# -----------------------------
# INPUT STATISTICS
# -----------------------------

print("\nRAW INPUT STATISTICS:")

for i, name in enumerate(channel_names):

    vals = Xtr[..., i][:, mask]
    vals = vals[~np.isnan(vals)]

    print(
        f"{name}: "
        f"mean={vals.mean():.6f}, "
        f"std={vals.std():.6f}, "
        f"min={vals.min():.6f}, "
        f"max={vals.max():.6f}"
    )


print("\n" + "=" * 60)
print("AUDIT COMPLETE")
print("=" * 60)
```
