```python
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

# Reload data
X_val_raw = np.load("X_val.npy")
y_val_raw = np.load("y_val.npy")
ocean_mask_raw = np.load("ocean_mask.npy")

mask = ocean_mask_raw.astype(bool)
if mask.ndim == 3:
    mask = mask[0]

# Reload scaler and standardize X_val the same way as training
scaler = np.load("input_scaler.npy")
means, stds = scaler[0], scaler[1]

def prep(X):
    Z = (X - means) / stds
    Z[:, ~mask, :] = 0.0
    return np.nan_to_num(Z, nan=0.0).astype(np.float32)

X_val_clean = prep(X_val_raw)
y_val_clean = np.nan_to_num(y_val_raw, nan=0.0).astype(np.float32)

# Rebuild dataset/loader
class OceanDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32).permute(0, 3, 1, 2)
        self.y = torch.tensor(y, dtype=torch.float32).permute(0, 3, 1, 2)
    def __len__(self): return len(self.X)
    def __getitem__(self, idx): return self.X[idx], self.y[idx]

val_dataset = OceanDataset(X_val_clean, y_val_clean)
val_loader = DataLoader(val_dataset, batch_size=8, shuffle=False)

# Rebuild model architecture (must match training exactly)
class OceanCNN(nn.Module):
    def __init__(self, in_channels=3, out_channels=14):
        super().__init__()
        self.enc1 = nn.Sequential(nn.Conv2d(in_channels, 32, 3, padding=1), nn.ReLU(), nn.Conv2d(32, 32, 3, padding=1), nn.ReLU())
        self.pool1 = nn.MaxPool2d(2)
        self.enc2 = nn.Sequential(nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(), nn.Conv2d(64, 64, 3, padding=1), nn.ReLU())
        self.pool2 = nn.MaxPool2d(2)
        self.bottleneck = nn.Sequential(nn.Conv2d(64, 128, 3, padding=1), nn.ReLU(), nn.Conv2d(128, 128, 3, padding=1), nn.ReLU())
        self.up2 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.dec2 = nn.Sequential(nn.Conv2d(128, 64, 3, padding=1), nn.ReLU(), nn.Conv2d(64, 64, 3, padding=1), nn.ReLU())
        self.up1 = nn.ConvTranspose2d(64, 32, 2, stride=2)
        self.dec1 = nn.Sequential(nn.Conv2d(64, 32, 3, padding=1), nn.ReLU(), nn.Conv2d(32, 32, 3, padding=1), nn.ReLU())
        self.out = nn.Conv2d(32, out_channels, 1)
    def forward(self, x):
        e1 = self.enc1(x); p1 = self.pool1(e1)
        e2 = self.enc2(p1); p2 = self.pool2(e2)
        b = self.bottleneck(p2)
        u2 = self.up2(b); d2 = self.dec2(torch.cat([u2, e2], dim=1))
        u1 = self.up1(d2); d1 = self.dec1(torch.cat([u1, e1], dim=1))
        return self.out(d1)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = OceanCNN(in_channels=3, out_channels=14).to(device)
model.load_state_dict(torch.load("best_model.pth", map_location=device))
model.eval()

# Run inference
all_preds = []
with torch.no_grad():
    for X_batch, y_batch in val_loader:
        X_batch = X_batch.to(device)
        pred = model(X_batch).cpu().numpy()
        all_preds.append(pred)

preds = np.concatenate(all_preds, axis=0).transpose(0, 2, 3, 1)  # (N_val, H, W, 14)

print("preds shape:", preds.shape)
print("mask shape:", mask.shape)

# Save for use in argo_validation.ipynb
np.save("val_predictions.npy", preds)
np.save("ocean_mask_bool.npy", mask)
print("Saved predictions and mask for ARGO validation.")
```

```python
import numpy as np
import xarray as xr
import pandas as pd

argo_oa = xr.open_dataset("argo_oa_validation.nc")

# Your model grid
lat_grid = np.arange(5.5, 24.26, 0.25)
lon_grid = np.arange(80.5, 99.26, 0.25)

# Regrid CORA to model's 0.25° grid
argo_regridded = argo_oa.interp(
    latitude=lat_grid,
    longitude=lon_grid,
    method="linear"
)

# Match model depths
standard_depths = [
    0.5, 5, 10, 20, 30, 50, 75,
    100, 125, 150, 200, 300, 500, 700
]

argo_aligned = argo_regridded.interp(
    depth=standard_depths,
    method="linear"
)

print("Aligned CORA shape:", argo_aligned.TEMP.shape)
print(
    "NaN fraction after regrid:",
    np.isnan(argo_aligned.TEMP.values).mean()
)
```

```python
# Load the clean predictions
preds_clean = np.load("val_predictions_maskfix_v1.npy")

val_dates = pd.date_range(
    "2020-05-25",
    "2020-06-30",
    freq="D"
)

pred_monthly = {}

for month in [5, 6]:
    mask = val_dates.month == month

    pred_monthly[month] = preds_clean[mask].mean(axis=0)

    print(
        f"Month {month}: "
        f"{mask.sum()} days averaged"
    )
```

```python
cora_months_dt = pd.to_datetime(argo_aligned.time.values)

for month in [5, 6]:

    cora_idx = np.where(
        cora_months_dt.month == month
    )[0][0]

    # CORA: (depth, lat, lon)
    cora_slice = argo_aligned.TEMP.values[cora_idx]

    # Convert to (lat, lon, depth)
    cora_slice = np.transpose(
        cora_slice,
        (1, 2, 0)
    )

    # Only compare where CORA has observations
    valid = np.isfinite(cora_slice)

    true_vals = cora_slice[valid]
    pred_vals = pred_monthly[month][valid]

    rmse = np.sqrt(
        np.mean((pred_vals - true_vals) ** 2)
    )

    mae = np.mean(
        np.abs(pred_vals - true_vals)
    )

    bias = np.mean(
        pred_vals - true_vals
    )

    corr = np.corrcoef(
        true_vals,
        pred_vals
    )[0, 1]

    print(f"\nMonth {month} vs CORA:")
    print(f"  Valid comparison points: {valid.sum()}")
    print(f"  RMSE: {rmse:.3f} °C")
    print(f"  MAE:  {mae:.3f} °C")
    print(f"  Bias: {bias:.3f} °C")
    print(f"  Corr: {corr:.3f}")
```

```python
depths = np.array([
    0.5, 5, 10, 20, 30, 50, 75,
    100, 125, 150, 200, 300, 500, 700
])

print("Depth-wise CORA validation:\n")

for i, depth in enumerate(depths):

    # CORA for this depth: (lat, lon)
    cora = argo_aligned.TEMP.values[:, i, :, :]

    # Convert model monthly predictions:
    # pred_monthly[month] is (lat, lon, depth)
    may_pred = pred_monthly[5][:, :, i]
    june_pred = pred_monthly[6][:, :, i]

    # May
    may_true = cora[4]       # May is index 4
    may_valid = np.isfinite(may_true)

    may_t = may_true[may_valid]
    may_p = may_pred[may_valid]

    may_rmse = np.sqrt(np.mean((may_p - may_t) ** 2))
    may_mae = np.mean(np.abs(may_p - may_t))
    may_bias = np.mean(may_p - may_t)
    may_corr = np.corrcoef(may_t, may_p)[0, 1]

    # June
    june_true = cora[5]      # June is index 5
    june_valid = np.isfinite(june_true)

    june_t = june_true[june_valid]
    june_p = june_pred[june_valid]

    june_rmse = np.sqrt(np.mean((june_p - june_t) ** 2))
    june_mae = np.mean(np.abs(june_p - june_t))
    june_bias = np.mean(june_p - june_t)
    june_corr = np.corrcoef(june_t, june_p)[0, 1]

    print(
        f"{depth:>5} m | "
        f"May RMSE={may_rmse:5.3f}, "
        f"MAE={may_mae:5.3f}, "
        f"Bias={may_bias:+5.3f}, "
        f"Corr={may_corr:5.3f} | "
        f"June RMSE={june_rmse:5.3f}, "
        f"MAE={june_mae:5.3f}, "
        f"Bias={june_bias:+5.3f}, "
        f"Corr={june_corr:5.3f}"
    )
```
